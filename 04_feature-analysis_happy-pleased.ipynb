```python
import pandas as pd
import numpy as np

import hvplot.pandas
import holoviews as hv
hv.extension('bokeh')

from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.inspection import permutation_importance
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_predict


from sklearn.metrics import f1_score
```

```python
df_train = pd.read_csv('emotions/emotions_train.csv')
```

```python
emo_cols = [
    'amazed-suprised',
    'happy-pleased',
    'relaxing-calm',
    'quiet-still',
    'sad-lonely',
    'angry-aggresive'
]

X_train = df_train.drop(columns=emo_cols)
y_train = df_train[emo_cols]
```

```python
# calculate mutual information for happy-pleased
target = y_train['happy-pleased']

mi_scores = mutual_info_classif(
    X_train,
    target, 
    random_state=42
) 

mi_df = pd.DataFrame({
    'feature': X_train.columns,
    'mutual_information': mi_scores
}).sort_values(
    'mutual_information',
    ascending=False
)

mi_df.head(20)
```

```python
rf_hp = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf_hp.fit(X_train, target)
```

```python
# RF feature importance for happy-pleased

rf_importance = pd.DataFrame({
    'feature': X_train.columns,
    'rf_importance': rf_hp.feature_importances_
}).sort_values(
    'rf_importance',
    ascending=False
)

rf_importance.head(20)
```

```python
# permutation importance using stratified cross validation due to the small size of the dataset
cv = StratifiedKFold(
    n_splits=5, 
    shuffle=True,
    random_state=42
)

all_importances = []

for train_idx, val_idx in cv.split(X_train, target):
    
    X_fold_train = X_train.iloc[train_idx]
    X_fold_val = X_train.iloc[val_idx]

    y_fold_train = target.iloc[train_idx]
    y_fold_val = target.iloc[val_idx]

    rf_fold = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )

    rf_fold.fit(X_fold_train, y_fold_train)

    perm = permutation_importance(
        rf_fold,
        X_fold_val,
        y_fold_val,
        scoring='f1',
        n_repeats=10,
        random_state=42,
        n_jobs=-1
    )

    all_importances.append(
        perm.importances_mean
    )
```

```python
# baseline
baseline_pred = rf_fold.predict(X_fold_val)
baseline_f1 = f1_score(y_fold_val, baseline_pred)

print("Baseline F1:", baseline_f1)
```

```python
# combining the folds

all_importances = np.array(all_importances)

perm_df = pd.DataFrame({
    'feature': X_train.columns,
    'importance_mean': all_importances.mean(axis=0),
    'importance_std':  all_importances.std(axis=0)
}).sort_values(
    'importance_mean',
    ascending=False
)

perm_df.head(20)
```

Mutual information found only weak individual dependencies between the available features and happy-pleased. Random Forest impurity importance did not identify a small set of dominant predictors, while cross-validated permutation importance showed small and unstable decreases in F1 when individual features were disrupted. Taken together, these findings suggest that the class is unlikely to be explained by a few strongly predictive individual features. Any useful signal may instead be distributed across multiple variables, shared among redundant features, or contained in feature interactions.

```python
# MI stability across folds

target = y_train['happy-pleased']

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

mi_fold_results = []

for fold, (train_idx, val_idx) in enumerate(cv.split(X_train, target), start=1):

    X_fold = X_train.iloc[train_idx]
    y_fold = target.iloc[train_idx]

    mi_scores = mutual_info_classif(
        X_fold,
        y_fold,
        random_state=42
    )

    fold_df = pd.DataFrame({
        'feature': X_train.columns,
        'mi': mi_scores,
        'fold': fold
    })

    mi_fold_results.append(fold_df)

mi_stability = pd.concat(
    mi_fold_results,
    ignore_index=True
)

mi_summary = (
    mi_stability
    .groupby('feature')
    .agg(
        mean_mi=('mi', 'mean'),
        std_mi=('mi', 'std')
    )
    .sort_values('mean_mi', ascending=False)
)

mi_summary.head(20)
```

```python
top_10_counts = (
    mi_stability
    .sort_values(['fold', 'mi'], ascending=[True, False])
    .groupby('fold')
    .head(10)
    ['feature']
    .value_counts()
)

mi_summary['top10_count'] = top_10_counts

mi_summary['top10_count'] = (
    mi_summary['top10_count']
    .fillna(0)
    .astype(int)
)

mi_summary.head(20)
```

```python
mi_rank = (
    mi_summary
    .reset_index()[['feature', 'mean_mi']]
)

mi_rank['mi_rank'] = (
    mi_rank['mean_mi']
    .rank(ascending=False)
)

rf_rank = rf_importance[
    ['feature', 'rf_importance']
].copy()

rf_rank['rf_rank'] = (
    rf_rank['rf_importance']
    .rank(ascending=False)
)

perm_rank = perm_df[
    ['feature', 'importance_mean']
].copy()

perm_rank['perm_rank'] = (
    perm_rank['importance_mean']
    .rank(ascending=False)
)
```

```python
ranking_comparison = (
    mi_rank
    .merge(rf_rank, on='feature')
    .merge(perm_rank, on='feature')
)

ranking_comparison['average_rank'] = (
    ranking_comparison[
        ['mi_rank', 'rf_rank', 'perm_rank']
    ]
    .mean(axis=1)
)

ranking_comparison = ranking_comparison.sort_values(
    'average_rank'
)

ranking_comparison.head(20)
```

The three feature-importance methods do not strongly agree on a small set of dominant features. Some MFCC-related features consistently appear relatively high, but their importance varies substantially depending on the method used.
