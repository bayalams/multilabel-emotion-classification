# Conclusions

## Overview

This project explored multilabel classification of musical emotions using 72 numerical audio features.

The main objective was not only to compare models, but also to understand why certain emotion labels were easier to classify than others and whether poor performance could be improved through model refinement.

---

## Logistic Regression Baseline

Logistic Regression with One-vs-Rest provided a useful baseline.

Because each emotion was modeled independently, this approach did not make use of possible relationships between emotion labels.

Despite its simplicity, Logistic Regression performed relatively well overall and achieved better F1-based performance than the independent Random Forest model.

For `happy-pleased`, Logistic Regression achieved approximately:

- Precision: 0.40
- Recall: 0.55
- F1-score: 0.46
- ROC-AUC: 0.63

The ROC-AUC of 0.63 suggested that the model had some ability to distinguish positive from negative samples, but the underlying separation was relatively weak.

This indicated that threshold tuning alone was unlikely to completely solve the problem for this model.

---

## Random Forest Baseline

Random Forest was introduced to test whether nonlinear relationships between the audio features and the emotion labels could improve classification.

The independent Random Forest model behaved differently from Logistic Regression.

It generally produced:

- higher precision;
- lower recall;
- fewer positive predictions.

This made the model more conservative.

For `happy-pleased`, the initial Random Forest model achieved a low recall and F1-score, despite having better precision.

However, its ROC-AUC was approximately 0.70.

This was an important result because it showed that the model was often ranking positive `happy-pleased` samples above negative samples even though relatively few crossed the default threshold required to be classified as positive.

---

## Effect of the Classifier Chain

A Random Forest Classifier Chain was then tested to include dependencies between emotion labels.

Unlike One-vs-Rest, the Classifier Chain allows earlier label predictions to become additional information for later classifiers.

This improved overall multilabel performance compared with the independent Random Forest model.

The results suggested that relationships between emotion labels contain useful predictive information.

However, `happy-pleased` remained one of the weakest individual labels.

---

## Investigation of `happy-pleased`

Because `happy-pleased` remained difficult across several models, additional feature-level analysis was performed.

The investigation included:

- feature distributions;
- correlations;
- Mutual Information;
- Random Forest feature importance;
- permutation importance;
- comparison of feature rankings.

No single feature emerged as a dominant predictor.

Mutual Information values were generally low, and permutation importance values were small and unstable across folds.

This suggests that the difficulty of predicting `happy-pleased` is probably not caused by the absence of one important feature.

Instead, the useful signal may be:

- distributed across several variables;
- relatively weak;
- redundant across correlated features;
- dependent on interactions between features;
- partly dependent on relationships with other emotion labels.

---

## Changing the Classifier Chain Order

The position of `happy-pleased` in the Classifier Chain was also investigated.

Initially, `happy-pleased` appeared early in the chain and therefore had access to only a small number of previous label predictions.

It was moved to the end of the chain so that it could use information from all preceding emotion predictions.

This produced a small improvement in its classification performance.

Its F1-score increased from approximately 0.21 to 0.28.

Its ROC-AUC also improved:

- Original order: approximately 0.709
- New order: approximately 0.734

Most of the other emotion labels remained largely stable.

This suggests that `happy-pleased` benefited from additional information provided by the other emotion labels.

---

## Threshold Tuning

The strongest improvement came from changing the classification threshold.

The default threshold of 0.5 caused the Random Forest model to predict relatively few `happy-pleased` positives.

However, the ROC-AUC indicated that the model had learned useful ranking information.

Cross-validated predicted probabilities were therefore used to test different thresholds.

As the threshold increased:

- precision increased;
- recall decreased;
- the F1-score changed depending on the balance between the two.

The threshold that maximized the F1-score produced a substantial improvement.

After tuning, the `happy-pleased` results were approximately:

- Precision: 0.51
- Recall: 0.61
- F1-score: 0.56

This was considerably better than the original Random Forest prediction behaviour.

The result demonstrates that part of the poor baseline performance was caused by the decision rule applied to the model's probabilities rather than by a complete lack of predictive information.

---

## Overall Interpretation

The project produced several useful conclusions.

First, model complexity alone did not guarantee better performance. Logistic Regression initially outperformed the independent Random Forest model on several F1-based metrics.

Second, multilabel dependencies mattered. The Classifier Chain improved Random Forest performance and showed that emotion labels contain useful information about one another.

Third, poor F1-score should not be interpreted in isolation. ROC-AUC showed that the Random Forest model had considerably more ranking ability for `happy-pleased` than its original recall and F1-score suggested.

Finally, threshold selection had a major effect on the practical behaviour of the classifier.

The best-performing decision process was therefore not obtained simply by replacing one algorithm with another, but through a sequence of diagnosis and targeted refinement.

---

## Limitations

The dataset is relatively small and contains 72 numerical features.

This makes results more sensitive to sampling variation and increases the risk of overfitting.

Detailed semantic descriptions of many audio features were also unavailable.

Because of this, domain-driven feature engineering was limited.

The feature analysis therefore focused primarily on statistical relationships and model-dependent importance measures.

Another limitation is that threshold tuning was primarily investigated for `happy-pleased`. The other labels may also benefit from individually optimized thresholds.

---

## Future Work

Several extensions could be investigated:

- tune classification thresholds separately for all emotion labels;
- compare several Classifier Chain orders;
- use ensembles of multiple Classifier Chains;
- perform model-specific feature selection;
- test dimensionality reduction;
- evaluate gradient boosting methods;
- evaluate Support Vector Machines;
- investigate feature interactions;
- revisit audio-specific feature engineering if additional feature documentation becomes available.

These experiments should continue to use cross-validation for model selection and preserve the test set for final evaluation.

---

## Final Conclusion

The main lesson from the project is that understanding model behaviour can be more valuable than simply adding increasingly complex models.

The `happy-pleased` label initially appeared to be poorly learned by Random Forest because its recall and F1-score were low.

Further analysis showed a more nuanced picture.

The model had moderate ranking ability, benefited from information from other labels through the Classifier Chain, and improved substantially when the decision threshold was adjusted.

This transformed the problem from:

> "The model cannot predict this emotion."

into:

> "The model has learned useful information, but the default decision rule does not use that information effectively."

That distinction became one of the most important findings of the project.
