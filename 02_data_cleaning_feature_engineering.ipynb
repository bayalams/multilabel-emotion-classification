```python
from scipy.io import arff
import pandas as pd
```

```python
data, met = arff.loadarff("emotions/emotions-train.arff")
```

```python
df_train = pd.DataFrame(data)
```

```python
df_train.info()
```

```python
emo_cols= [
    'amazed-suprised',
    'happy-pleased',
    'relaxing-calm',
    'quiet-still',
    'sad-lonely',
    'angry-aggresive'
]
```

```python
#change type of byte variables to string - Yes or No
for col in emo_cols:
    df_train[col] = df_train[col].str.decode('utf-8')

    df_train[col] = df_train[col].map(
        {'0': '0',
        '1': '1'}
    )
```
