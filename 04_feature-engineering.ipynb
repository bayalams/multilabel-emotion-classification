```python
import pandas as pd
import numpy as np

import hvplot.pandas
import holoviews as hv
hv.extension('bokeh')

from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.inspection import permutation_importance
```

```python
df_train = pd.read_csv('emotions/emotions_train.csv')
```

```python
emo_cols = [
    'amazed-suprised',
    'happy-pleased',
    'relaxing-calm',
    'quiet-still',
    'sad-lonely',
    'angry-aggresive'
]

X_train = df_train.drop(columns=emo_cols)
y_train = df_train[emo_cols]
```

```python
# calculate mutual information for happy-pleased
target = y_train['happy-pleased']

mi_scores = mutual_info_classif(
    X_train,
    target, 
    random_state=42
) 

mi_df = pd.DataFrame({
    'feature': X_train.columns,
    'mutual_information': mi_scores
}).sort_values(
    'mutual_information',
    ascending=False
)

mi_df.head(20)
```

```python
top_mi = (
    mi_df
    .head(20)
    .sort_values('mutual_information')
)

top_mi.hvplot.bar(
    x='feature',
    y='mutual_information',
    invert=True,
    title='Top Mutual Information Scores — happy-pleased',
    xlabel='Mutual Information',
    ylabel='Feature',
    width=900,
    height=600
)
```

```python
feature_1 = mi_df.iloc[0]['feature']
feature_2 = mi_df.iloc[1]['feature']

plot_df = pd.DataFrame({
    feature_1: X_train[feature_1],
    feature_2: X_train[feature_2],
    'happy_pleased': y_train['happy-pleased'].astype(str)
})

plot_df.hvplot.scatter(
    x=feature_1,
    y=feature_2,
    by='happy_pleased',
    alpha=0.6,
    width=700,
    height=500,
    title='Top MI Features — happy-pleased'
)
```

```python
rf_hp = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf_hp.fit(X_train, target)
```

```python
# RF feature importance for happy-pleased

rf_importance = pd.DataFrame({
    'feature': X_train.columns,
    'rf_importance': rf_hp.feature_importances_
}).sort_values(
    'rf_importance',
    ascending=False
)

rf_importance.head(20)
```

```python
# permutation importance using stratified cross validation due to the small size of the dataset

cv = StratifiedFold(
    n_splits=5, 
    shuffle=True,
    random_state=42
)

    
perm = permutation_importance(
    rf_hp,
    X_train,
    target,
    scoring='f1',
    n_repeats=20,
    random_state=42,
    n_jobs=-1
)
   
```
